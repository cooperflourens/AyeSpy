# Set up tokenizer

We're gonna use the HuggingFace framework to load a model and embed some ish

```python
# This model is small, but effective. We can play with other models if we find this one isn't very good.
model_id = "sentence-transformers/all-MiniLM-L6-v2"
# Can make a HF account and pass in your token here, find it at http://hf.co/settings/tokens
hf_token = ""
```

```python
# Rather than load the model here (since we're just using a pretrained one) we'll just hit the hosted one in HF
api_url = f"https://api-inference.huggingface.co/pipeline/feature-extraction/{model_id}"
headers = {"Authorization": f"Bearer {hf_token}"}
```

# Tokenize

```python
def query(texts):
    response = requests.post(api_url, headers=headers, json={"inputs": texts, "options":{"wait_for_model":True}})
    return response.json()
```

```python
embeddings = query(documents)
```
